# Summarising Data

Once you are happy that your data are clean and in the shape you want it (wide or long), you'll usually want to summarise it and produce some descriptive statistics. If your data are in wide format, using `summary()` is a straightforward way to get a fair bit of useful information.

```r
summary(messyDat)
```

```text
    pcp_id             pcp_age       congruent_rt   incongruent_rt  pcp_edu_level completed_exp 
 Length:7           Min.   :18.00   Min.   :1.770   Min.   :2.890   School:3      Mode:logical  
 Class :character   1st Qu.:18.25   1st Qu.:1.995   1st Qu.:3.055   UG    :2      TRUE:7        
 Mode  :character   Median :19.50   Median :2.100   Median :3.405   PG    :2                    
                    Mean   :19.50   Mean   :2.206   Mean   :3.477                               
                    3rd Qu.:20.00   3rd Qu.:2.290   3rd Qu.:3.860                               
                    Max.   :22.00   Max.   :3.000   Max.   :4.210                               
                    NA's   :1                       NA's   :1  
```

This tells us quite a lot including min, max, mean and NAs. Because we cleaned these data we also have useful summaries of education and age, which we didn't when we first imported it. In Psychology we generally also want to know standard deviations. We can get these ourselves fairly easily though. 

```r
sd_rt <- sapply(
  X = messyDat[c("congruent_rt", "incongruent_rt","pcp_age")],
  FUN = sd,
  na.rm = TRUE
)
```
This code calculates the standard deviation for the `congruent_rt`, `incongruent_rt` and `age` columns, ignoring missing values.

* `messyDat[c("congruent_rt", "incongruent_rt","pcp_age")]` extracts those three columns from the data frame.
* `sapply()` applies the same function to each selected column.
* `X = ` supplies the data to work through. 
* `FUN = sd` tells R to calculate the standard deviation.

The result is stored in `sd_rt`, a named numeric vector:

```r
sd_rt
```
```text
  congruent_rt incongruent_rt        pcp_age 
     0.4099129      0.5311748      1.5165751 
```

We could have obtained each of those values using the simpler `sd(messyDat$congruent_rt, na.rm = TRUE)`, repeated once for each variable. But if you had 20 columns, `sapply()` would handle all of them in the same three lines, and swapping in a different function for `FUN` makes it just as easy to calculate the mean, median, or any other statistic across those columns.

## Summarising Long Data

In long-format data, each participant usually has more than one row: one row for each condition, time point, or measurement. This means that `summary()` and `sapply()` calculate statistics for the whole `reaction_time` column rather than separately for each condition.

To calculate a separate mean for each condition, use `aggregate()`.
```r
aggregate(reaction_time ~ condition,data=messyDatLong,FUN = mean,na.rm=T)
```
```text
    condition reaction_time
1   congruent      2.205714
2 incongruent      3.476667
```
In `aggregate()`, the variable on the left of the tilde (`~`) is the outcome being summarised, and the variable on the right is the grouping variable.

The formula `reaction_time ~ condition` can be read as:

> Calculate a summary of `reaction_time`, separately for each level of `condition`.

* `reaction_time` is the numeric variable to summarise.
* `condition` is the grouping variable.
* `data = messyDatLong` identifies the data frame containing these variables.
* `FUN = mean` tells R to calculate the mean for each group.


This is even more useful when you have several independent variables in your data, as we did with our `exp` data earlier:

```r
aggregate(accuracy ~ cue_type + difficulty,data=exp,FUN = mean,na.rm=T)
```
```text
  cue_type difficulty accuracy
1    audio       easy     83.0
2   visual       easy     93.5
3    audio       hard     80.0
4   visual       hard     71.5
```
Now we have a mean for all four conditions in the experiment. `accuracy ~ cue_type + difficulty` is saying summarise accuracy for every level in cue type and difficulty.
