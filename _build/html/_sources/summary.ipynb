# Summary

We started by importing data. Using `read.csv()`, you can load data from a local machine or directly from a URL. Arguments like `header`, and `na.strings` give us precise control over how R reads a file.

We then inspected the data, because before we clean or analyse anything, we need to know what we're working with. Functions like `str()`, and `summary()`, give us an overview of a dataset's structure, variable types, and any obvious irregularities like missing values, or variables stored as the wrong type.

In the cleaning section we practised correcting variable types (converting a character column to numeric with `as.numeric()`, encoding an ordinal variable as a factor with `factor()`), handling missing values with `is.na()`, and removing duplicate rows using `!duplicated()`.

In reshaping data we covered the difference between wide and long format, and why long format is often required in R. Using `reshape()`, we converted a wide dataset into long format.

In practice, data rarely arrives clean and ready to use. It needs importing, checking, correcting, and restructuring before any summarising or inference can be done. Getting comfortable with this workflow now will save a great deal of time later.
