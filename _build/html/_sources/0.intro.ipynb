# Introduction: Importing Wrangling & Summarising

This week, we will develop essential skills for preparing and understanding data before analysis. Real datasets are rarely ready to analyse immediately: they may contain missing values, inconsistent categories, or require restructuring.

We will begin by importing, inspecting, and cleaning data in R. We will then explore factors, which are used to represent categorical variables such as experiment conditions. Next, we will reshape data between wide and long formats. Finally, we will calculate useful summaries, including and group-level descriptive statistics.
