# Summary

This week, we covered several fundamental techniques for extracting and filtering data in R.

We began by selecting values from vectors using their index, or position. We learned how to select individual values, multiple values, ranges of values, and values to exclude using negative indexing.

We then introduced logical conditions and saw how comparison operators such as `<`, `>`, `==`, `<=`, and `>=` produce logical vectors containing `TRUE` and `FALSE` values. These logical vectors can be used as filters, or masks, to select values that meet particular criteria. We also used logical operators such as `&`, `|`, and `!` to combine or reverse conditions.

We discussed missing values and saw that comparisons involving `NA` produce `NA`. We learned how to identify missing values using `is.na()` and how to exclude them explicitly when filtering data.

We then introduced data frames, which are two-dimensional data structures consisting of rows and columns. We learned how to inspect their dimensions, column names, and structure using functions such as `nrow()`, `ncol()`, `dim()`, `names()`, and `str()`.

Finally, we learned how to select rows and columns from data frames using square brackets and how to select named columns using the `$` operator. We used logical conditions to subset data frames and introduced the `subset()` function as a convenient alternative for filtering rows and selecting columns.

These techniques form the foundation for working with datasets in R. They allow us to identify, extract, and prepare the observations and variables needed for further analysis.
